the problem with normal server client API interactions is that when we write the code to fetch the server data through APIs we have no control over the server side API calls... this means we have to change our code whenever there is a change in the server side change in the api url, api internal workings, or return format of the api 

we have to write different code for every connection we make with different api services

model context protocol - mcp

in this case the server side before the api hit the server there is an mcp layer which is built which handles these kinds of changes 

this mcp server consist of functions for different api 

we can directly access the data through the mcp server and do not worry about the server side code changes

we can also use a json file and directly connect to multiple mcp servers at the same time so that we can avoid writing redundant codes

MCP is an open source standard for connecting AI applications to external systems 

MCP architecture 

there are three main components of MCP

MCP host - the AI application that coordinates and manages one or multiple MCP clients

MCP client - a component that maintains a connection to an MCP server and obtain the context from the MCP server for the MCP host to use it 

MCP server  - a program which provides the context to the mcp clients 

there are two types of MCP servers 

these are classified based on the transport layers used by the MCP servers 

transport layer is how the MCP servers and clients communicate 

stdio is used when we download the entire MCP server inside our machine and use this 
streamable Http is when we use the http version of the mcp server directly without downloading ti locally

coroutine is a function which can take asynchronous routine 

we use mcp inspector while inspecting streamable-http transportable servers
